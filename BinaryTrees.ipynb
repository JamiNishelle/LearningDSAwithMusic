```python
# Binary Tree Implementation
class Node:
  def __init__(self, val):
    self.val = val
    self.left = None
    self.right = None
```

```python
# example tree to traverse

root = Node(1)
root.left = Node(2)
root.right = Node(3)
root.left.left = Node(4)
root.left.right = Node(5)
```

```python
def inorder(node):
    if node is None:
        return
    else:
        inorder(node.left)
        print(node.val)
        inorder(node.right)
```

```python
inorder(root)
```

```python
def preorder(node):
    if node is None:
        return
    else:
        print(node.val)
        preorder(node.left)
        preorder(node.right)
```

```python
preorder(root)
```

```python
def postorder(node):
    if node is None:
        return
    else:
        postorder(node.left)
        postorder(node.right)
        print(node.val)
```

```python
postorder(root)
```

```python
# Breadth-first search
from collections import deque

def bfs(root):
    queue = deque() # deque is the class — you need to create an instance of it
    queue.append(root)
    while len(queue) > 0: # while the queue is not empty, i.e. has elements in it
        child = queue.popleft()
        print(child.val)
        if child.left is not None:
            queue.append(child.left)
        if child.right is not None:
            queue.append(child.right)
        
```

```python
bfs(root)
```

```python
# collect the values from each traversal in a list instead of printing them
# use the list as my melody
def inorder_list(node):
    if node is None:
        return []
    else:
        left_values = inorder_list(node.left)
        right_values = inorder_list(node.right)
        return left_values + [node.val] + right_values
    
def preorder_list(node):
    if node is None:
        return []
    else:
        left_values = preorder_list(node.left)
        right_values = preorder_list(node.right)    
        return [node.val] + left_values + right_values
       
 
def postorder_list(node):
    if node is None:
        return []
    else:
        left_values = postorder_list(node.left)
        right_values = postorder_list(node.right)
        return left_values + right_values + [node.val]
    
def bfs_list(root):
    queue = deque() # deque is the class — you need to create an instance of it
    queue.append(root)
    result = [] # creating an empty list 
    while len(queue) > 0: # while the queue is not empty, i.e. has elements in it
        child = queue.popleft()
        result.append(child.val)
        if child.left is not None:
            queue.append(child.left)
        if child.right is not None:
            queue.append(child.right)
    return result
```

```python
melodies = [inorder_list(root), preorder_list(root), postorder_list(root), bfs_list(root)]

for i, melody in enumerate(melodies):
    # Create a new Stream for this melody
    score = stream.Stream()
    
    # Convert MIDI numbers to Note objects and add to stream
    for num in melody:
        n = note.Note(num)
        n.quarterLength = 1.0  # Set duration (1.0 = quarter note)
        score.append(n)
    
    # Write to a MIDI file
    score.write('midi', fp=f'binarytree_{i}.mid')
    print(f'Saved binarytree_{i}.mid with {len(melody)} notes')
```

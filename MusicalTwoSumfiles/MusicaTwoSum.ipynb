# Two Sum - Pair with given Sum

```python
# Approach 1: Brute Force 
# time O(n^2) because of the 2 loops Space O(1) because we aren't storing anything.

def two_sum_approach1(array, target):
    len_array = len(array)
    for i in range(len_array):
        for j in range(i+1, len_array): # need this to start the inner loop at i+1, you want range(i+1, len_array)
            if target == array[i]+ array[j]:
                return True # no else statement is necessary because the loop automatically continues
    return False

# 	O(n²) — for each of the n elements, the inner loop does up to n steps, giving n×n in the worst case
#	O(1) — you're only using a fixed number of variables regardless of input size, so no extra memory needed
```

```python
# Approach 2: Sort + Binary Search  

def binary_search(arr, needle):
    left = 0
    right = len(arr) - 1 # Subtract one because index starts at zero
    while left <= right: # we allow for left and right to equal the same element
        mid = (left + right) // 2
        if arr[mid] == needle:
            return [True, mid]
        elif needle < arr[mid]:
            # search left half
            right = mid - 1 # Needle is in the left half, so shrink from the right: right = mid - 1
        else:
            # search right half
            left = mid + 1 # Needle is in the right half, so shrink from the left: left = mid + 1
    return [False, None] # None is Python's way of representing "no value"

def two_sum_approach2(array, target):
    sorted_array = sorted(array) # sorted() is Python's built-in function

    for i in range(len(sorted_array)):
        [value, index] = binary_search(sorted_array, target - sorted_array[i])
        if value:
            if i != index:
                return True
    return False

```

```python
arr1 = [0, -1, 2, -3, 1]
print(two_sum_approach2(arr1, -2))  # Expected: True
```

```python
# Approach 3 — Sort + Two Pointers

def two_sum_approach3(array, target):
    sorted_array = sorted(array) # sorted() is Python's built-in function
    len_array = len(sorted_array)
    left = 0
    right = len_array - 1
    while left < right:
        if sorted_array[left] + sorted_array[right] == target:
            return True
        elif sorted_array[left] + sorted_array[right] < target: # too small so move left one to the right
            left += 1
        elif sorted_array[left] + sorted_array[right] > target: # too big so move right one to the left
            right -= 1

    return False

# Yay, this works!
```

```python
print(two_sum_approach3(arr1, -2))
print(two_sum_approach3(arr2, 0)) 
```

```python
# Approach 4 — the hash set 

def two_sum_approach4(array, target):
    check_set = set()
    for i in range(len(array)):
        if target - array[i] in check_set:
            return True
        else:
            check_set.add(array[i]) # add the element we saw to the hash set
    return False
```

```python
print(two_sum_approach4(arr1, -2))
print(two_sum_approach4(arr2, 0)) 
```
